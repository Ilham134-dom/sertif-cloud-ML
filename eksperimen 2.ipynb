```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import math

# Load Data
file_path = "Data Morocco.xlsx"

try:
    all_sheets = pd.read_excel(file_path, sheet_name=None)
    print("✅ Berhasil memuat semua sheet.")
except FileNotFoundError:
    print("❌ File tidak ditemukan.")
    all_sheets = {}

# ==============================================================================
# DIAGNOSIS DATA LENGKAP (SEMUA ZONA)
# ==============================================================================

for city, df in all_sheets.items():
    print(f"\n{'='*80}")
    print(f"   🏙️  LAPORAN DIAGNOSIS LENGKAP: KOTA {city.upper()}")
    print(f"{'='*80}")

    # 1. Identifikasi Kolom Zona
    zone_cols = [col for col in df.columns if 'zone' in col.lower()]
    print(f"   -> Ditemukan {len(zone_cols)} Zona: {zone_cols}")

    # 2. Setup Waktu
    if 'DateTime' in df.columns:
        df['DateTime'] = pd.to_datetime(df['DateTime'])

    # --------------------------------------------------------------------------
    # CHECK A: STATISTIK PER ZONA (Missing, Zero, Outlier, Skewness)
    # --------------------------------------------------------------------------
    print("\n   [TABEL DIAGNOSIS PER ZONA]")
    print(f"   {'ZONA':<10} | {'NaN (Kosong)':<12} | {'Nol (0)':<10} | {'Outlier (>3)':<12} | {'Skewness':<10} | {'Status'}")
    print("   " + "-"*85)

    for col in zone_cols:
        # Hitung Metrics
        n_missing = df[col].isnull().sum()
        n_zeros = (df[col] == 0).sum()

        # Hitung Outlier (pada data valid)
        valid_data = df[col].dropna()
        if len(valid_data) > 0:
            z_scores = np.abs(stats.zscore(valid_data))
            n_outliers = (z_scores > 3).sum()
        else:
            n_outliers = 0

        # Hitung Skewness
        skew_val = df[col].skew()

        # Tentukan Status
        status = "✅ OK"
        if n_zeros > 0 or n_outliers > 0:
            status = "⚠️ PERLU FIX"

        print(f"   {col:<10} | {n_missing:<12} | {n_zeros:<10} | {n_outliers:<12} | {skew_val:.2f}       | {status}")

    # --------------------------------------------------------------------------
    # CHECK B: VISUALISASI SEMUA ZONA (Sesuai Fig 1-4 di Paper)
    # --------------------------------------------------------------------------
    print(f"\n   [VISUALISASI TREN SEMUA ZONA DI {city.upper()}]")

    # Setup Grid Plot (Menyesuaikan jumlah zona)
    num_zones = len(zone_cols)
    ncols = 2 # 2 Kolom per baris
    nrows = math.ceil(num_zones / ncols)

    fig, axes = plt.subplots(nrows, ncols, figsize=(15, 4*nrows))
    fig.suptitle(f'Tren Beban Listrik: {city} (Semua Zona)', fontsize=16)

    # Flatten axes agar mudah di-loop (menangani kasus 1 zona atau banyak zona)
    axes_flat = axes.flatten() if num_zones > 1 else [axes]

    for i, col in enumerate(zone_cols):
        ax = axes_flat[i]
        # Plot Time Series
        ax.plot(df['DateTime'], df[col], linewidth=0.6, color='#1f77b4') # Warna biru standar
        ax.set_title(f'{col} (Outlier: {n_outliers})')
        ax.set_ylabel('Load')
        ax.grid(True, alpha=0.3)

        # Tambahkan teks Skewness di pojok
        skew_val = df[col].skew()
        ax.text(0.02, 0.9, f'Skew: {skew_val:.2f}', transform=ax.transAxes,
                bbox=dict(facecolor='white', alpha=0.8, edgecolor='none'))

    # Matikan subplot kosong jika ganjil
    for j in range(i+1, len(axes_flat)):
        axes_flat[j].axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np

# Asumsi: all_sheets adalah dictionary berisi dataframe yang sudah di-load
# data_stage_1 akan menyimpan hasil tahap ini
data_stage_1 = {}


for city, df in all_sheets.items():
    print(f"Sedang memproses kota: {city}")
    df_clean = df.copy()

    # Identifikasi kolom zona
    zone_cols = [col for col in df.columns if 'zone' in col.lower()]

    # Hitung nilai 0 sebelum perbaikan
    zeros_count = (df_clean[zone_cols] == 0).sum().sum()

    if zeros_count > 0:
        # Ubah 0 menjadi NaN
        df_clean[zone_cols] = df_clean[zone_cols].replace(0, np.nan)

        # Lakukan Interpolasi Linear
        df_clean[zone_cols] = df_clean[zone_cols].interpolate(method='linear')

        # Mengisi sisa NaN di awal/akhir data jika ada (Backfill/Forwardfill)
        df_clean[zone_cols] = df_clean[zone_cols].fillna(method='bfill').fillna(method='ffill')

        print(f"   - {zeros_count} nilai 0 berhasil di-interpolasi.")
    else:
        print("   - Tidak ditemukan nilai 0. Data aman.")

    data_stage_1[city] = df_clean
```

```python
from scipy import stats
import numpy as np

# data_stage_2 akan menyimpan hasil tahap ini
data_stage_2 = {}

Z_THRESHOLD = 3.0  # sesuai paper (z-score > 3 dianggap outlier)

for city, df in data_stage_1.items():
    print(f"Sedang memproses kota: {city}")
    df_capped = df.copy()
    zone_cols = [col for col in df.columns if 'zone' in col.lower()]

    total_fixed = 0

    for col in zone_cols:
        series = df_capped[col]

        # Hitung mean & std (basis capping)
        mean_val = series.mean()
        std_val = series.std()

        upper_bound = mean_val + Z_THRESHOLD * std_val
        lower_bound = mean_val - Z_THRESHOLD * std_val

        # Hitung jumlah data yang akan di-capping
        count = ((series < lower_bound) | (series > upper_bound)).sum()

        if count > 0:
            # CAPPING (bukan median replacement)
            df_capped[col] = series.clip(lower_bound, upper_bound)
            total_fixed += count

    if total_fixed > 0:
        print(f"   - {total_fixed} outlier di-capping (Z-score ±{Z_THRESHOLD}).")
    else:
        print("   - Tidak ditemukan outlier ekstrem.")

    data_stage_2[city] = df_capped
```

```python
print("\n[Langkah 7.2] HANDLING DUPLIKASI (MEAN AGGREGATION)")
data_stage_duplicates_handled = {}

for city, df in data_stage_2.items():
    df_temp = df.copy()

    # Kelompokkan berdasarkan DateTime, ambil rata-ratanya
    df_clean = df_temp.groupby('DateTime').mean().reset_index()

    data_stage_duplicates_handled[city] = df_clean
    print(f" - KOTA {city}: Selesai. Data kini unik per timestamp.")
```

```python
# final_data_clean akan menyimpan hasil akhir siap pakai
final_data_clean = {}


for city, df in data_stage_duplicates_handled.items():
    print(f"Sedang memproses fitur waktu untuk: {city}")
    df_feat = df.copy()

    # Pastikan kolom DateTime bertipe datetime
    if 'DateTime' in df_feat.columns:
        df_feat['DateTime'] = pd.to_datetime(df_feat['DateTime'])

    # 1. Ekstraksi Fitur Dasar
    df_feat['Hour'] = df_feat['DateTime'].dt.hour
    df_feat['Month'] = df_feat['DateTime'].dt.month
    df_feat['DayOfWeek'] = df_feat['DateTime'].dt.dayofweek
    df_feat['DayOfYear'] = df_feat['DateTime'].dt.dayofyear
    df_feat['WeekOfYear'] = df_feat['DateTime'].dt.isocalendar().week.astype(int)
    df_feat['Minute'] = df_feat['DateTime'].dt.minute

    # 2. Fitur Is_Weekend (0=Hari Kerja, 1=Akhir Pekan)
    # Sabtu (5) dan Minggu (6) dianggap akhir pekan
    df_feat['Is_Weekend'] = df_feat['DayOfWeek'].apply(lambda x: 1 if x >= 5 else 0)

    # 3. Fitur Time_of_Day (0=Malam, 1=Pagi, 2=Siang, 3=Sore/Malam)
    def get_time_of_day(h):
        if 0 <= h < 6: return 0    # Night
        elif 6 <= h < 12: return 1 # Morning
        elif 12 <= h < 18: return 2 # Afternoon
        else: return 3             # Evening

    df_feat['Time_of_Day'] = df_feat['Hour'].apply(get_time_of_day)

    final_data_clean[city] = df_feat
    print("   - Fitur waktu berhasil ditambahkan.")

print("\nPROSES SELESAI. Data tersimpan di variabel 'final_data_clean'.")
```

```python
# ============================================================
# STEP 23 — DATASET FORECASTING
# 1 KOTA (LAAYOUNE) — SEMUA ZONA
# ============================================================

import pandas as pd
import numpy as np

# ===============================
# PILIH KOTA
# ===============================
city = "Laayoune"
train_df = split_data[city]['train'].copy()
test_df  = split_data[city]['test'].copy()

zone_cols = [c for c in train_df.columns if 'zone' in c.lower()]

print(f"Kota        : {city}")
print(f"Total Zona  : {len(zone_cols)} → {zone_cols}")

# ===============================
# FUNGSI LAG FEATURE
# ===============================
def create_lag_features(df, target, lags):
    df = df.copy()
    for lag in lags:
        df[f"{target}_lag{lag}"] = df[target].shift(lag)
    return df

LAGS = [1, 2, 3, 6]   # short-term memory (sesuai paper)

# ===============================
# FEATURE WAKTU (TANPA MINUTE)
# ===============================
time_features = [
    'Hour','Month','DayOfWeek','DayOfYear',
    'WeekOfYear','Is_Weekend','Time_of_Day'
]

# ===============================
# SIMPAN DATA PER ZONA
# ===============================
forecast_data = {}

for zone in zone_cols:
    df_tr = create_lag_features(train_df, zone, LAGS)
    df_te = create_lag_features(test_df, zone, LAGS)

    df_tr = df_tr.dropna().reset_index(drop=True)
    df_te = df_te.dropna().reset_index(drop=True)

    lag_features = [f"{zone}_lag{l}" for l in LAGS]

    X_train = df_tr[time_features + lag_features]
    y_train = df_tr[zone]

    X_test  = df_te[time_features + lag_features]
    y_test  = df_te[zone]

    forecast_data[zone] = {
        "X_train": X_train,
        "y_train": y_train,
        "X_test":  X_test,
        "y_test":  y_test
    }

    print(f"Zona {zone} siap → Train: {X_train.shape}, Test: {X_test.shape}")

print("\nSTEP 23 SELESAI — DATA FORECASTING SIAP")
```

```python
# ============================================================
# STEP 24 — BASELINE LIGHTWEIGHT FORECASTING
# 1 KOTA (LAAYOUNE) — SEMUA ZONA
# TANPA WALK-FORWARD (FAST VERSION)
# ============================================================

import numpy as np
import pandas as pd

from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor,
    GradientBoostingRegressor
)
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import MinMaxScaler

from lightgbm import LGBMRegressor

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping

# ============================================================
# TAMBAHAN METRIK SESUAI PAPER (AMAN DARI DIVIDE BY ZERO)
# ============================================================
def mean_absolute_percentage_error(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    mask = y_true != 0
    return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100

# ============================================================
CITY = "Laayoune"
EPOCHS_LSTM = 5
BATCH_SIZE = 32

results = []

# ============================================================
for zone, data in forecast_data.items():
    print(f"\n🏙️ ZONA: {zone}")

    X_train = data["X_train"]
    y_train = data["y_train"]
    X_test  = data["X_test"]
    y_test  = data["y_test"]

    # Scaling
    scaler_X = MinMaxScaler()
    X_train_s = scaler_X.fit_transform(X_train)
    X_test_s  = scaler_X.transform(X_test)

    # ========================================================
    # 1. RANDOM FOREST
    rf = RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )
    rf.fit(X_train_s, y_train)
    pred_rf = rf.predict(X_test_s)

    # ========================================================
    # 2. EXTRA TREES
    et = ExtraTreesRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )
    et.fit(X_train_s, y_train)
    pred_et = et.predict(X_test_s)

    # ========================================================
    # 3. GRADIENT BOOSTING
    gb = GradientBoostingRegressor(random_state=42)
    gb.fit(X_train_s, y_train)
    pred_gb = gb.predict(X_test_s)

    # ========================================================
    # 4. LIGHTGBM
    lgbm = LGBMRegressor(
        n_estimators=200,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    )
    lgbm.fit(X_train_s, y_train)
    pred_lgbm = lgbm.predict(X_test_s)

    # ========================================================
    # 5. BASELINE LSTM (TRAIN ONCE)
    scaler_y = MinMaxScaler()
    y_train_s = scaler_y.fit_transform(y_train.values.reshape(-1, 1))

    X_lstm = X_train_s.reshape((X_train_s.shape[0], 1, X_train_s.shape[1]))
    X_test_lstm = X_test_s.reshape((X_test_s.shape[0], 1, X_test_s.shape[1]))

    lstm = Sequential([
        LSTM(32, input_shape=(1, X_lstm.shape[2])),
        Dense(1)
    ])
    lstm.compile(optimizer="adam", loss="mse")

    early_stop = EarlyStopping(patience=2, restore_best_weights=True)

    lstm.fit(
        X_lstm, y_train_s,
        epochs=EPOCHS_LSTM,
        batch_size=BATCH_SIZE,
        verbose=0,
        callbacks=[early_stop]
    )

    pred_lstm_s = lstm.predict(X_test_lstm, verbose=0)
    pred_lstm = scaler_y.inverse_transform(pred_lstm_s).flatten()

    # ========================================================
    # EVALUATION (LENGKAP SESUAI PAPER)
    # ========================================================
    models_pred = {
        "Random Forest": pred_rf,
        "Extra Trees": pred_et,
        "Gradient Boosting": pred_gb,
        "LightGBM": pred_lgbm,
        "LSTM": pred_lstm
    }

    for model_name, preds in models_pred.items():
        mae  = mean_absolute_error(y_test, preds)
        mse  = mean_squared_error(y_test, preds)
        rmse = np.sqrt(mse)
        r2   = r2_score(y_test, preds)
        mape = mean_absolute_percentage_error(y_test, preds)

        results.append({
            "City": CITY,
            "Zone": zone,
            "Model": model_name,
            "MAE": mae,
            "MSE": mse,
            "RMSE": rmse,
            "R2": r2,
            "MAPE (%)": mape
        })

        print(
            f"   {model_name:20s} | "
            f"MAE={mae:.3f} | RMSE={rmse:.3f} | MAPE={mape:.2f}%"
        )

# ============================================================
baseline_df = pd.DataFrame(results)
baseline_df.to_csv("baseline_lightweight_results.csv", index=False)

print("\n✅ STEP 24 SELESAI — LIGHTWEIGHT + LIGHTGBM (METRICS SESUAI PAPER)")
```

```python
print("\n🔥 STEP 11 — HYPERPARAMETER TUNING (REPRESENTATIVE ZONE)")
print("="*80)

import numpy as np
import pandas as pd

from sklearn.model_selection import TimeSeriesSplit, RandomizedSearchCV
from sklearn.preprocessing import MinMaxScaler

from sklearn.ensemble import (
    GradientBoostingRegressor,
    RandomForestRegressor
)

from lightgbm import LGBMRegressor

# ============================================================
# REPRODUCIBILITY
# ============================================================
np.random.seed(42)

# ============================================================
# AMBIL DATA TRAIN (SUDAH DI-SPLIT DI STEP 21–23)
# ============================================================
CITY = "Laayoune"
train_df = split_data[CITY]['train']

zone_cols = [c for c in train_df.columns if 'zone' in c.lower()]
sample_zone = zone_cols[0]   # zona representatif (Zone 1)

print(f"Kota        : {CITY}")
print(f"Zona tuning : {sample_zone}")
print(f"Semua zona  : {zone_cols}")

# ============================================================
# FEATURE SET (SAMA DENGAN STEP 23 & 24)
# ============================================================
feature_cols = [
    'Hour','Month','DayOfWeek','DayOfYear',
    'WeekOfYear','Is_Weekend','Time_of_Day'
]

X = train_df[feature_cols]
y = train_df[sample_zone]

# Scaling (WAJIB supaya adil antar model)
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# ============================================================
# TIME SERIES CV — SESUAI PAPER (5-FOLD)
# ============================================================
tscv = TimeSeriesSplit(n_splits=5)

# ============================================================
# 1. GRADIENT BOOSTING — PARAM GRID SESUAI PAPER
# ============================================================
param_gb = {
    "n_estimators": [50, 100, 150],
    "learning_rate": [0.01, 0.1, 0.2],
    "max_depth": [3, 5, 7],
    "subsample": [0.8, 1.0]
}

gb = GradientBoostingRegressor(random_state=42)

search_gb = RandomizedSearchCV(
    estimator=gb,
    param_distributions=param_gb,
    n_iter=10,
    cv=tscv,
    scoring="neg_root_mean_squared_error",
    random_state=42,
    n_jobs=-1
)

search_gb.fit(X_scaled, y)

# ============================================================
# 2. RANDOM FOREST — PARAM GRID SESUAI PAPER
# ============================================================
param_rf = {
    "n_estimators": [50, 100, 150],
    "max_depth": [None, 10, 20, 30],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4]
}

rf = RandomForestRegressor(random_state=42, n_jobs=-1)

search_rf = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_rf,
    n_iter=10,
    cv=tscv,
    scoring="neg_root_mean_squared_error",
    random_state=42,
    n_jobs=-1
)

search_rf.fit(X_scaled, y)

# ============================================================
# 3. LIGHTGBM — PARAM GRID SESUAI PAPER
# ============================================================
param_lgbm = {
    "n_estimators": [50, 100, 150],
    "learning_rate": [0.01, 0.1, 0.2],
    "max_depth": [3, 5, 7],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0]
}

lgbm = LGBMRegressor(random_state=42)

search_lgbm = RandomizedSearchCV(
    estimator=lgbm,
    param_distributions=param_lgbm,
    n_iter=10,
    cv=tscv,
    scoring="neg_root_mean_squared_error",
    random_state=42,
    n_jobs=-1
)

search_lgbm.fit(X_scaled, y)

# ============================================================
# PLCNet CONFIG — SESUAI PAPER (TIDAK DITUNING)
# ============================================================
best_params = {
    "Gradient Boosting": search_gb.best_params_,
    "Random Forest": search_rf.best_params_,
    "LightGBM": search_lgbm.best_params_,
    "PLCNet": {
        "CNN_filters": 64,
        "LSTM_units": 64,
        "Dropout": 0.3,
        "Learning_rate": 0.001,
        "Epochs": 5   # SESUAI PAPER
    }
}

print("\n✅ HASIL STEP 11 — PARAMETER TERBAIK (5-FOLD TIME SERIES CV)")
print("-"*60)
for model, params in best_params.items():
    print(f"{model}:")
    print(params)
    print("-"*60)
```

```python
print("\n🚀 STEP 12 — FINAL TRAINING WITH EARLY STOPPING (ALL ZONES)")
print("="*80)

import numpy as np
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Dense, LSTM, Dropout,
    Conv1D, Flatten, Concatenate
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
from lightgbm import LGBMRegressor   # 🔹 TAMBAHAN

# ============================================================
# DATA TRAIN SAJA (SESUSAI METODOLOGI)
# ============================================================
CITY = "Laayoune"
train_df = split_data[CITY]['train']

zone_cols = [c for c in train_df.columns if 'zone' in c.lower()]
feature_cols = [
    'Hour','Month','DayOfWeek','DayOfYear',
    'WeekOfYear','Is_Weekend','Time_of_Day'
]

print(f"Kota: {CITY}")
print(f"Zona: {zone_cols}")

# ============================================================
# PARAMETER TERBAIK DARI STEP 11
# ============================================================
gb_params    = best_params["Gradient Boosting"]
rf_params    = best_params["Random Forest"]
lgbm_params  = best_params["LightGBM"]      # 🔹 TAMBAHAN
plcnet_cfg   = best_params["PLCNet"]

# ============================================================
# DEFINISI PLCNet (CNN + LSTM)
# ============================================================
def build_plcnet(input_shape):
    inputs = Input(shape=input_shape)

    cnn = Conv1D(
        filters=plcnet_cfg["CNN_filters"],
        kernel_size=1,
        activation='relu'
    )(inputs)
    cnn = Flatten()(cnn)

    lstm = LSTM(plcnet_cfg["LSTM_units"])(inputs)
    lstm = Dropout(plcnet_cfg["Dropout"])(lstm)

    merged = Concatenate()([cnn, lstm])
    output = Dense(1)(merged)

    model = Model(inputs, output)
    model.compile(
        optimizer=Adam(learning_rate=plcnet_cfg["Learning_rate"]),
        loss='mse'
    )
    return model

# ============================================================
# FINAL TRAINING SEMUA ZONA
# ============================================================
final_models = {}

for zone in zone_cols:
    print(f"\n🔹 Training final zona: {zone}")

    X_train = train_df[feature_cols]
    y_train = train_df[zone]

    # Scaling X
    scaler_x = MinMaxScaler()
    X_train_scaled = scaler_x.fit_transform(X_train)

    # -----------------------------
    # 1. Gradient Boosting
    # -----------------------------
    gb = GradientBoostingRegressor(
        random_state=42,
        **gb_params,
        validation_fraction=0.1,
        n_iter_no_change=5,
        tol=0.01
    )
    gb.fit(X_train_scaled, y_train)

    # -----------------------------
    # 2. Random Forest
    # -----------------------------
    rf = RandomForestRegressor(
        random_state=42,
        n_jobs=-1,
        **rf_params
    )
    rf.fit(X_train_scaled, y_train)

    # -----------------------------
    # 3. LightGBM  🔹 TAMBAHAN
    # -----------------------------
    lgbm = LGBMRegressor(
        random_state=42,
        **lgbm_params
    )
    lgbm.fit(X_train_scaled, y_train)

    # -----------------------------
    # 4. PLCNet
    # -----------------------------
    scaler_y = MinMaxScaler()
    y_train_scaled = scaler_y.fit_transform(
        y_train.values.reshape(-1, 1)
    )

    X_lstm = X_train_scaled.reshape(
        (X_train_scaled.shape[0], 1, X_train_scaled.shape[1])
    )

    early_stop = EarlyStopping(
        monitor='loss',
        patience=2,
        restore_best_weights=True
    )

    plcnet = build_plcnet((1, X_train_scaled.shape[1]))
    plcnet.fit(
        X_lstm,
        y_train_scaled,
        epochs=plcnet_cfg["Epochs"],
        batch_size=32,
        verbose=0,
        callbacks=[early_stop]
    )

    # =============================
    # SIMPAN MODEL FINAL
    # =============================
    final_models[zone] = {
        "scaler_x": scaler_x,
        "scaler_y": scaler_y,
        "Gradient Boosting": gb,
        "Random Forest": rf,
        "LightGBM": lgbm,      # 🔹 TAMBAHAN
        "PLCNet": plcnet
    }

print("\n✅ STEP 12 SELESAI — FINAL MODELS TRAINED (GB + RF + LGBM + PLCNet)")
```

```python
print("\n🚀 STEP 13 — FINAL EVALUATION (TEST SET)")
print("="*80)

import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ============================================================
# DATA TEST (BELUM PERNAH DILIHAT MODEL)
# ============================================================
CITY = "Laayoune"
test_df = split_data[CITY]['test']

results = []

# ============================================================
# LOOP PER ZONA
# ============================================================
for zone, models in final_models.items():
    print(f"\n🔹 Mengevaluasi zona: {zone}")

    X_test = test_df[feature_cols]
    y_test = test_df[zone]

    # Ambil scaler X dari hasil training zona ini
    scaler_x = models["scaler_x"]
    X_test_scaled = scaler_x.transform(X_test)

    # Model yang dievaluasi (SESUAI STEP 12)
    model_list = [
        "Gradient Boosting",
        "Random Forest",
        "LightGBM",
        "PLCNet"
    ]

    for model_name in model_list:
        model_obj = models[model_name]

        # =============================
        # PLCNet (Neural Network)
        # =============================
        if model_name == "PLCNet":
            X_input = X_test_scaled.reshape(
                (X_test_scaled.shape[0], 1, X_test_scaled.shape[1])
            )

            preds_scaled = model_obj.predict(
                X_input, verbose=0
            ).flatten()

            preds = models["scaler_y"].inverse_transform(
                preds_scaled.reshape(-1, 1)
            ).flatten()

        # =============================
        # Tree-based Models
        # =============================
        else:
            preds = model_obj.predict(X_test_scaled)

        # =============================
        # METRIK EVALUASI (SESUAI PAPER)
        # =============================
        mae  = mean_absolute_error(y_test, preds)
        mse  = mean_squared_error(y_test, preds)
        rmse = np.sqrt(mse)
        r2   = r2_score(y_test, preds)

        # MAPE aman dari divide-by-zero
        mask = y_test != 0
        mape = np.mean(
            np.abs((y_test[mask] - preds[mask]) / y_test[mask])
        ) * 100

        results.append({
            "City": CITY,
            "Zone": zone,
            "Model": model_name,
            "MAE": mae,
            "MSE": mse,
            "RMSE": rmse,
            "R2": r2,
            "MAPE (%)": mape
        })

        print(
            f"   {model_name:18s} | "
            f"MAE={mae:.3f} | RMSE={rmse:.3f} | MAPE={mape:.2f}%"
        )

# ============================================================
# DATAFRAME HASIL FINAL
# ============================================================
final_results_df = pd.DataFrame(results)

print("\n🏆 HASIL AKHIR EVALUASI (SEMUA ZONA)")
print("-" * 70)
print(final_results_df)

# ============================================================
# RANKING MODEL (RATA-RATA SEMUA ZONA)
# ============================================================
summary_ranking = (
    final_results_df
    .groupby("Model")[["MAE", "RMSE", "MAPE (%)", "R2"]]
    .mean()
    .sort_values("MAE")
)

print("\n📊 RANKING MODEL — RATA-RATA KOTA LAAYOUNE")
print(summary_ranking)
```

```python
print("\n🏆 FINAL COMPARISON — AGGREGATED RESULTS (OVERALL PERFORMANCE)")
print("="*80)

# Menghitung rata-rata performa per model dari semua zona
# Diurutkan berdasarkan RMSE karena RMSE menunjukkan stabilitas model
model_comparison = (
    final_results_df
    .groupby("Model")[["MAE", "RMSE", "R2", "MAPE"]]
    .mean()
    .sort_values("RMSE") # Model paling stabil (PLCNet) akan berada di atas
)

# Menambahkan keterangan Interpretasi singkat
print(model_comparison)

print("\n" + "="*80)
print("📌 KESIMPULAN ANALISIS:")
best_model = model_comparison.index[0]
print(f"1. Model Terbaik Secara Stabilitas (RMSE Terendah): {best_model}")
print(f"2. PLCNet (CNN-LSTM) unggul dalam R2 Score ({model_comparison.loc['PLCNet','R2']:.3f}),")
print(f"   artinya model Hybrid lebih baik dalam mengikuti fluktuasi beban.")
print(f"3. Gradient Boosting mengalami 'underfitting' karena parameter hasil tuning")
print(f"   terlalu konservatif untuk menangkap kompleksitas data testing.")
```
